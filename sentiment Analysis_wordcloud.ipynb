```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob # natural language meaning
from wordcloud import WordCloud    # create the wordcloud
```

```python
#load the file
data=pd.read_csv("Sentiment_Analysis .csv")
data.head(10)
```

```python
#define the sentiment analysis function

def get_sentiment(text):
    
    meaning =TextBlob(text)

    if meaning.sentiment.polarity >0:
        return "Positive"
        
    elif meaning.sentiment.polarity <0:
        return "Negative" 
        
    else: 
         return "Neutral"

        
```

```python
# Apply sentiment Analysis

data["Sentiment"] = data["Sentences"].apply(get_sentiment)
```

```python
# display the data

data.head(10)
```

```python
data.to_excel("Avi_sentiment_update.xlsx",index=False)
```

```python
data["Sentiment"].value_counts()
```

```python
#filter out all the negative reviews

negative_reviews=data[data["Sentiment"]=="Negative"]
negative_reviews.head(9)
```

```python
# combine all negative sentences in one paragraph

all_negative_text=" ".join(negative_reviews["Sentences"])
print(all_negative_text)
```

```python
# create a wordcloud for negative reviews

diagram = WordCloud(width=800, height=400, background_color="white",colormap="rainbow").generate(all_negative_text)
```

```python
#plot the wordcloud

plt.figure(figsize=(10,5))
plt.imshow(diagram,interpolation="bilinear")
plt.title("Most common words in Negative Feedbacks")
plt.axis="off"
plt.show()
```
